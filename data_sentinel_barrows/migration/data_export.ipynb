```python
# Notebook: export_data.py
# Run on SOURCE workspace

catalog = "learning_dev"
schema  = "s819534"
export_base = f"/Volumes/{catalog}/{schema}/exports/data_sentinel/data"

tables = [
    # Registry/config tables — small, export as parquet
    "uc1_table_registry_b",
    "uc1_column_registry_b",
    "uc1_column_validation_registry_b",
    "uc1_table_validation_registry_b",
    "uc1_baseline_metrics_b",
    "uc2_layer_registry_b",
    "uc2_layer_check_config_b",
    "uc2_reconciliation_metric_config_b",
    "uc3_transformation_registry_b",
    "uc3_review_results_b",
    "uc4_kpi_registry_b",
    "uc4_kpi_baseline_metrics_b",
    # Run/results tables — partitioned, larger
    "uc1_validation_run_b",
    "uc1_validation_results_b",
    "uc2_reconciliation_run_b",
    "uc2_reconciliation_results_b",
    "uc2_silver_store_transactions",
    "uc2_silver_inventory_movements",
    "uc2_gold_daily_store_metrics",
    "uc2_gold_daily_inventory_summary",
    "uc4_kpi_validation_run_b",
    "uc4_kpi_validation_results_b",
]

for table in tables:
    fqn         = f"{catalog}.{schema}.{table}"
    export_path = f"{export_base}/{table}"
    try:
        df = spark.table(fqn)
        df.write.mode("overwrite").parquet(export_path)
        count = df.count()
        print(f"✅ {table}: {count:,} rows → {export_path}")
    except Exception as e:
        print(f"❌ {table}: {e}")

print("\nExport complete.")
print(f"Location: {export_base}")
```
